3. RSA Overview
- Metaheuristic algorithm inspired by crocodile hunting behavior
- Two phases: Exploration and Exploitation
- Goal: Select the most discriminative subset of features

4. RSA Implementation
Implement or import RSA with fitness function based on classification performance.

4. RSA Implementation (Update ver.)
RSA was implemented to select the most discriminative features from the 32-dimensional latent space. 

### Key Improvements from Baseline Version
- Increased population size to 40 and iterations to 50 for better exploration
- Strengthened fitness evaluation (Random Forest with 80 estimators, 5-fold CV) on only the Training Set
- Enhanced Exploitation: 45% probability of following the current best solution
- Enhanced Exploration: 12% mutation rate + 5% random reset for diversity
- Elitism: Always preserve the global best solution across generations

### RSA Parameters
- Population size: 40
- Maximum iterations: 50
- Number of features: 32
- Fitness function: 5-fold CV accuracy using Random Forest

```python
class ReptileSearchAlgorithm:
    def __init__(self, n_features, population_size=40, max_iter=50):
        self.n_features = n_features
        self.pop_size = population_size
        self.max_iter = max_iter
        self.best_fitness = -np.inf
        self.best_solution = None
    
    def fitness(self, solution):
        selected_idx = np.where(solution == 1)[0]
        if len(selected_idx) < 3:   # 최소 3개 특징은 선택되도록
            return 0.0
        
        X_sel = X_train_latent[:, selected_idx]
        
        clf = RandomForestClassifier(
            n_estimators=80,      # 50 -> 80 
            max_depth=12,
            random_state=42,
        )
        
        scores = cross_val_score(
            clf, X_sel, y_train,
            cv=5,                 # 3-> 5
            scoring='accuracy'
        )
        return scores.mean()
    
    def optimize(self):
        # 초기 population
        population = np.random.randint(0, 2, size=(self.pop_size, self.n_features))
        
        for iteration in tqdm(range(self.max_iter), desc="RSA Optimization"):
            for i in range(self.pop_size):
                fit = self.fitness(population[i])
                if fit > self.best_fitness:
                    self.best_fitness = fit
                    self.best_solution = population[i].copy()
            
            population[0] = self.best_solution.copy()               # Elitism: 최적해 보존
            
            # Population update logic
            for i in range(1, self.pop_size):
                if np.random.rand() < 0.45:
                    mask = np.random.rand(self.n_features) < 0.5
                    population[i][mask] = self.best_solution[mask]
                
                if np.random.rand() < 0.12:
                    m_idx = np.random.randint(self.n_features)
                    population[i][m_idx] = 1 - population[i][m_idx]
                
                if np.random.rand() < 0.05:
                    population[i] = np.random.randint(0, 2, self.n_features)
        
        selected_count = int(self.best_solution.sum()) if self.best_solution is not None else 0
        print(f"\n[Final Results]")
        print(f"Best Accuracy (CV): {self.best_fitness:.4f}")
        print(f"Selected Features: {selected_count} / 32")
        return self.best_solution

# Run
rsa = ReptileSearchAlgorithm(n_features=32, population_size=40, max_iter=50)
best_solution = rsa.optimize()
```
